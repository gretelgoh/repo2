```python
# Inter-rater reliability

from sklearn.metrics import cohen_kappa_score
import numpy as np
import krippendorff

df = pd.read_csv("/content/dataset_train_test.csv")

# Cohen's Kappa
r1 = df["rater1"]
r2 = df["rater2"]

#correct for non integer in rater 1
r1 = pd.to_numeric(df["rater1"], errors="coerce").astype(int)
r2 = pd.to_numeric(df["rater2"], errors="coerce").astype(int)

kappa     = cohen_kappa_score(r1, r2)
agreement = (r1 == r2).mean()

# Encode string labels to numbers for krippendorff
encode     = {label: i for i, label in enumerate(r1.unique())}
r1_encoded = r1.map(encode).tolist()
r2_encoded = r2.map(encode).tolist()

# krippendorff expects shape (raters, items)
alpha = krippendorff.alpha(
    reliability_data=np.array([r1_encoded, r2_encoded]),
    level_of_measurement="nominal"
)

# ── Results ───────────────────────────────────────────────────
print(f"\nCohen's Kappa:        {kappa:.4f}")
print(f"Krippendorff's Alpha: {alpha:.4f}")
print(f"Percent Agreement:    {agreement:.2%}")
```
