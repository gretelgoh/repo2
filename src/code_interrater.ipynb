```python
# Inter Rater Reliability

# Inter-rater reliability
!pip install krippendorff

import pandas as pd
import numpy as np
from sklearn.metrics import cohen_kappa_score, confusion_matrix
import krippendorff

df = pd.read_csv("/content/dataset_train_test.csv")

# -------------------------------------------------
# 2. Cohen's kappa (rater1 vs rater2 only)
# -------------------------------------------------

r1 = df["rater1"].astype(int)
r2 = df["rater2"].astype(int)

kappa = cohen_kappa_score(r1, r2)
print(f"\n=== Cohen's kappa (rater1 vs rater2) ===")
print(f"κ = {kappa:.4f}")
print(f"N = {len(r1)}")

# Confusion matrix
print("\nConfusion matrix (rows=rater1, cols=rater2):")
print(confusion_matrix(r1, r2))

# -------------------------------------------------
# 3. Krippendorff's alpha
# -------------------------------------------------
reliability_data = np.array([
    df["rater1"].values,   # rater 1
    df["rater2"].values,   # rater 2
    df["rater3"].values    # rater 3 (mostly NaN)
], dtype=float)


alpha = krippendorff.alpha(
    reliability_data=reliability_data,
    level_of_measurement="nominal"   # change if your labels are ordinal
)

print(f"\n=== Krippendorff's alpha ===")
print(f"α = {alpha:.4f}")

# -------------------------------------------------
# Agreement
# -------------------------------------------------
print("\n=== Agreement diagnostics ===")
agree = (df["rater1"] == df["rater2"])
print(f"Agreement (rater1 == rater2): {agree.mean():.1%}  ({agree.sum()} / {len(df)})")




# Third rater used
third_used = df["rater3"].notna()
print(f"Rows that needed rater3: {third_used.sum()}  ({third_used.mean():.1%})")

# Confusion matrix Visualization
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

# Generate confusion matrix from your original data
cm = confusion_matrix(r1, r2, labels=[0, 1])

# Plot heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    cbar=True,
    xticklabels=["0", "1"],
    yticklabels=["0", "1"],
    annot_kws={"size": 16}
)

plt.title("Confusion Matrix: Rater 1 vs. Rater 2")
plt.xlabel("Rater 2")
plt.ylabel("Rater 1")
plt.tight_layout()
plt.show()
```
